```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# BASE DE DADOS

dados = {
    "Mes": ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"],
    "Notebook": [20, 25, 22, 30, 35, 40],
    "Mouse": [50, 55, 48, 60, 65, 70],
    "Teclado": [30, 28, 35, 40, 38, 45]
}

df = pd.DataFrame(dados)

print("BASE DE DADOS:")
print(df)
```

```python
# 1. EXPLORAÇÃO

print("\n1. PRIMEIRAS LINHAS:")
print(df.head())

print("\nDIMENSÕES DO DATAFRAME:")
print(df.shape)

print("\nESTATÍSTICAS DESCRITIVAS:")
print(df.describe())
```

```python
# 2. INDICADORES COM NUMPY

produtos = ["Notebook", "Mouse", "Teclado"]

print("\n2. INDICADORES DE VENDAS:")

for produto in produtos:
    vendas = df[produto].to_numpy()

    media = np.mean(vendas)
    maior = np.max(vendas)
    menor = np.min(vendas)

    print(f"\n{produto}:")
    print(f"Média: {media:.2f}")
    print(f"Maior venda: {maior}")
    print(f"Menor venda: {menor}")
```

```python
# 3. EVOLUÇÃO TEMPORAL

plt.figure(figsize=(9, 5))

for produto in produtos:
    plt.plot(df["Mes"], df[produto],
             marker="o", label=produto)

plt.title("Evolução das Vendas por Produto")
plt.xlabel("Meses")
plt.ylabel("Quantidade Vendida")
plt.legend()
plt.grid()
plt.show()
```

```python
# 4. COMPARAÇÃO

totais = df[produtos].sum()

plt.figure(figsize=(8, 5))
totais.plot(kind="bar")

plt.title("Total de Vendas por Produto")
plt.xlabel("Produtos")
plt.ylabel("Quantidade Total Vendida")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()
```

```python
# 5. DISTRIBUIÇÃO

produto_escolhido = "Notebook"

plt.figure(figsize=(8, 5))

plt.hist(df[produto_escolhido],
         bins=5, edgecolor="black")

plt.title(f"Distribuição das Vendas de {produto_escolhido}")
plt.xlabel("Quantidade Vendida")
plt.ylabel("Frequência")
plt.grid(axis="y")
plt.show()
```

```python
# 6. INTERPRETAÇÃO

print("\n6. INTERPRETAÇÃO DOS RESULTADOS:")

produto_maior = totais.idxmax()

print(f"\nProduto com maior volume total: {produto_maior}")
print(f"Total vendido: {totais.max()} unidades")

for produto in produtos:
    mes_maior = df.loc[df[produto].idxmax(), "Mes"]
    valor_maior = df[produto].max()

    print(f"{produto}: maior venda em {mes_maior}, "
          f"com {valor_maior} unidades.")
```

```python
# 7. NOVA VISUALIZAÇÃO

produto_escolhido = "Notebook"

media_notebook = df[produto_escolhido].mean()

cores = np.where(
    df[produto_escolhido] < media_notebook,
    "red",
    "green"
)

plt.figure(figsize=(9, 5))

plt.bar(df["Mes"], df[produto_escolhido], color=cores)

plt.axhline(media_notebook, color="blue",
            linestyle="--", label="Média de vendas")

plt.title("Vendas de Notebook em Relação à Média")
plt.xlabel("Meses")
plt.ylabel("Quantidade Vendida")
plt.legend()
plt.grid(axis="y")
plt.show()

print("\nANÁLISE ADICIONAL:")
print(f"Média de vendas de notebooks: {media_notebook:.2f}")

meses_abaixo = df.loc[
    df[produto_escolhido] < media_notebook, "Mes"
].tolist()

print("Meses abaixo da média:", meses_abaixo)
```
