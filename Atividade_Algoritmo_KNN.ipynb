```python
# Vai buscar o dataset

import pandas as pd

df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv',
                 skiprows = 0,
                 sep = ";")

pd.set_option('display.width', 166)

print(f'\n {df}')

print("\nPrimeiras linhas do dataset:")
print(df.head())
```

```python
def classificar_qualidade(nota):
  if nota <= 4:
    return "Baixa"
  elif nota <= 6:
    return "Média"
  else:
    return "Alta"

df["categoria"] = df["quality"].apply(classificar_qualidade)

print("\nDataset com as classificações:")
print(df)
```

```python
# X será a matriz de características

X = df[df.columns.difference(['quality', 'categoria'])].values

print(f'\nX: {X}')

# y será o vetor com a classificação

y = df['categoria'].values

print(f'\nY: {y}')
```

```python
from sklearn.neighbors import KNeighborsClassifier

vinho_classificador = KNeighborsClassifier(n_neighbors = 3)

vinho_classificador.fit(X, y)

print("\nModelo treinado com sucesso!")
```

```python
# validação

from sklearn.model_selection import cross_val_score

# Quebra o dataset em 5 partes

scores_dt = cross_val_score(vinho_classificador, X, y, scoring='accuracy', cv=5)

# média dos resultados

print(f'\nAcurácia: {scores_dt.mean()}')
```

```python
# Alterar o numero de vizinhos para uma acurácia melhor

for i in range(1,11):
  vinho_classificador = KNeighborsClassifier(n_neighbors = i)

  scores_dt = cross_val_score(vinho_classificador, X, y, scoring='accuracy', cv=5)

  print(f'Vizinhos: {i}  Acurácia: {scores_dt.mean()}')
```

```python
# Utilizando 3 vizinhos como exemplo

vinho_classificador = KNeighborsClassifier(n_neighbors = 3)

vinho_classificador.fit(X, y)

print("\nModelo final treinado com sucesso!")
```
