```python
import numpy as np

dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

dias = np.array([
    "Segunda", "Terça", "Quarta", "Quinta",
    "Sexta", "Sábado", "Domingo"
])


# PARTE 1


print("PARTE 1 — CONHECENDO OS DADOS")

print("Formato do array:", dados.shape)
print("Quantidade de linhas:", dados.shape[0])
print("Quantidade de colunas:", dados.shape[1])
print("Quantidade total de valores:", dados.size)
print("Número de dimensões:", dados.ndim)
print("Tipo dos dados:", dados.dtype)
```

```python
# PARTE 2

print("\nPARTE 2 — ACESSANDO OS DADOS")

print("2.1 - Sexta-feira:", dados[4])

print("2.2 - Faturamentos:", dados[:, 1])

print("2.3 - Reclamações:", dados[:, 4])

print("2.4 - Quinta-feira até domingo:\n", dados[3:7])

print("2.5 - Pedidos, faturamento e clientes:\n", dados[:, 0:3])
```

```python
# PARTE 3

print("\nPARTE 3 — OPERAÇÕES NUMÉRICAS")

total_pedidos = np.sum(dados[:, 0])
faturamento_total = np.sum(dados[:, 1])
total_clientes = np.sum(dados[:, 2])

media_pedidos = np.mean(dados[:, 0])
media_faturamento = np.mean(dados[:, 1])
media_reclamacoes = np.mean(dados[:, 4])

print("3.1 - Total de pedidos:", total_pedidos)
print("3.2 - Faturamento total: R$", faturamento_total)
print("3.3 - Total de clientes registrados:", total_clientes)
print("3.4 - Média diária de pedidos:", media_pedidos)
print("3.5 - Média diária de faturamento: R$", media_faturamento)
print("3.6 - Média de reclamações por dia:", media_reclamacoes)
```

```python
# PARTE 4

print("\nPARTE 4 — ANÁLISE POR DIA")

indice_mais_pedidos = np.argmax(dados[:, 0])
indice_menos_pedidos = np.argmin(dados[:, 0])

print("4.1 - Mais pedidos:", dias[indice_mais_pedidos])
print("Quantidade:", dados[indice_mais_pedidos, 0])

print("Menos pedidos:", dias[indice_menos_pedidos])
print("Quantidade:", dados[indice_menos_pedidos, 0])

indice_maior_faturamento = np.argmax(dados[:, 1])
indice_menor_faturamento = np.argmin(dados[:, 1])

print("4.2 - Maior faturamento:", dias[indice_maior_faturamento])
print("Valor: R$", dados[indice_maior_faturamento, 1])

print("Menor faturamento:", dias[indice_menor_faturamento])
print("Valor: R$", dados[indice_menor_faturamento, 1])

maior_faturamento = np.max(dados[:, 1])
menor_faturamento = np.min(dados[:, 1])
amplitude_faturamento = maior_faturamento - menor_faturamento

print("4.3 - Amplitude do faturamento: R$", amplitude_faturamento)
```

```python
# PARTE 5

print("\nPARTE 5 — MÁSCARAS E FILTROS")

mascara_faturamento = dados[:, 1] > 12000
print("5.1 - Dias com faturamento > R$ 12.000:", dias[mascara_faturamento])

mascara_pedidos = dados[:, 0] > 150
print("5.2 - Dias com mais de 150 pedidos:", dias[mascara_pedidos])

mascara_reclamacoes = dados[:, 4] > 4
print("5.3 - Dias com mais de 4 reclamações:", dias[mascara_reclamacoes])

print("5.4 - Faturamentos correspondentes:", dados[mascara_pedidos, 1])

mascara_abaixo_media = dados[:, 1] < media_faturamento
quantidade_abaixo_media = np.sum(mascara_abaixo_media)

print("5.5 - Dias abaixo da média:", dias[mascara_abaixo_media])
print("Quantidade de dias:", quantidade_abaixo_media)
```

```python
# PARTE 6

print("\nPARTE 6 — CLASSIFICAÇÃO DOS DIAS")

classificacao = np.where(
    dados[:, 1] >= 20000,
    "Alto desempenho",
    "Desempenho normal"
)

print("6.1 - Classificação:")
print(np.column_stack((dias, classificacao)))

quantidade_alto = np.sum(classificacao == "Alto desempenho")
quantidade_normal = np.sum(classificacao == "Desempenho normal")

print("Dias de alto desempenho:", quantidade_alto)
print("Dias de desempenho normal:", quantidade_normal)
```

```python
# PARTE 7

print("\nPARTE 7 — ESTATÍSTICA")

faturamentos = dados[:, 1]

media = np.mean(faturamentos)
mediana = np.median(faturamentos)
desvio_padrao = np.std(faturamentos)
minimo = np.min(faturamentos)
maximo = np.max(faturamentos)

print("Média: R$", media)
print("Mediana: R$", mediana)
print("Desvio padrão populacional: R$", desvio_padrao)
print("Mínimo: R$", minimo)
print("Máximo: R$", maximo)

print("Amplitude: R$", maximo - minimo)
print("Coeficiente de variação (%):", desvio_padrao / media * 100)
```

```python
# PARTE 8

print("\nPARTE 8 — TRABALHANDO COM AXIS")

medias_indicadores = np.mean(dados, axis=0)
maiores_indicadores = np.max(dados, axis=0)
menores_indicadores = np.min(dados, axis=0)
desvios_indicadores = np.std(dados, axis=0)

nomes_indicadores = np.array([
    "Pedidos", "Faturamento", "Clientes",
    "Ticket médio", "Reclamações"
])

print("8.1 - Média de cada indicador:")
print(np.column_stack((nomes_indicadores, medias_indicadores)))

print("8.2 - Maior valor de cada indicador:")
print(np.column_stack((nomes_indicadores, maiores_indicadores)))

print("8.3 - Menor valor de cada indicador:")
print(np.column_stack((nomes_indicadores, menores_indicadores)))

coeficientes_variacao = desvios_indicadores / medias_indicadores * 100
indice_maior_variacao = np.argmax(coeficientes_variacao)

print("8.4 - Coeficientes de variação (%):")
print(np.column_stack((nomes_indicadores, coeficientes_variacao)))
print("Maior variação relativa:", nomes_indicadores[indice_maior_variacao])

```

```python
# PARTE 9

print("\nPARTE 9 — ANÁLISE INTEGRADA")

melhor_dia = dias[indice_maior_faturamento]
print("9.1 - Melhor dia pelo faturamento:", melhor_dia)

dia_com_mais_pedidos = dias[indice_mais_pedidos]
print("9.2 - Também teve mais pedidos?",
      indice_maior_faturamento == indice_mais_pedidos)

indice_mais_reclamacoes = np.argmax(dados[:, 4])
print("9.3 - Também teve mais reclamações?",
      indice_maior_faturamento == indice_mais_reclamacoes)

mascara_integrada = (dados[:, 0] > 150) & (dados[:, 1] > 20000)

print("9.4 - Dias que atendem aos dois critérios:", dias[mascara_integrada])
print("Quantidade:", np.sum(mascara_integrada))

correlacao_pedidos_faturamento = np.corrcoef(
    dados[:, 0], dados[:, 1]
)[0, 1]

correlacao_pedidos_reclamacoes = np.corrcoef(
    dados[:, 0], dados[:, 4]
)[0, 1]

print("Correlação pedidos x faturamento:", correlacao_pedidos_faturamento)
print("Correlação pedidos x reclamações:", correlacao_pedidos_reclamacoes)
```

```python
# PARTE 10

print("\n" + "=" * 45)
print("RESUMO DA SEMANA")
print("=" * 45)

print("Total de pedidos:", total_pedidos)
print(f"Faturamento total: R$ {faturamento_total:,.2f}")
print(f"Média diária de pedidos: {media_pedidos:.2f}")
print(f"Média diária de faturamento: R$ {media_faturamento:,.2f}")
print("Melhor dia:", melhor_dia)
print("Pior dia:", dias[indice_menor_faturamento])
print("Maior número de reclamações:", np.max(dados[:, 4]))
print("Dias de alto desempenho:", dias[dados[:, 1] >= 20000])
```

```python
# PARTE 11

print("\nCONCLUSÃO")

print(
    f"A empresa registrou {total_pedidos} pedidos e faturou "
    f"R$ {faturamento_total:,.2f} durante a semana, com média diária "
    f"de R$ {media_faturamento:,.2f}."
)

print(
    f"O melhor dia em faturamento foi {melhor_dia}, com "
    f"R$ {maior_faturamento:,.2f}. O pior foi "
    f"{dias[indice_menor_faturamento]}, com R$ {menor_faturamento:,.2f}."
)

print(
    f"O desvio padrão do faturamento foi de R$ {desvio_padrao:,.2f}, "
    f"equivalente a {desvio_padrao / media * 100:.2f}% da média, "
    "indicando a variação relativa observada na semana."
)

print(
    f"O indicador com maior variação relativa foi "
    f"{nomes_indicadores[indice_maior_variacao]}, com "
    f"{coeficientes_variacao[indice_maior_variacao]:.2f}% de coeficiente "
    "de variação."
)

print(
    f"A correlação entre pedidos e faturamento foi "
    f"{correlacao_pedidos_faturamento:.2f}; entre pedidos e reclamações, "
    f"foi {correlacao_pedidos_reclamacoes:.2f}. Esses valores indicam "
    "associações nos dados observados, mas não comprovam causalidade."
)

print(
    f"{quantidade_abaixo_media} dias ficaram abaixo da média semanal "
    "de faturamento. Vale investigar os dias de menor resultado e "
    "acompanhar as reclamações nos dias de maior movimento."
)
```
