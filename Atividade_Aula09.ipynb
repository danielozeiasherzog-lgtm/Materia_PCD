```python
#Usandos os métodos do Pandas, localize a sua cidade de residência e:
# - mostre todas as informações
# - qual o mês de temperatura mínima
# - qual o mês de maior déficit hídrico
# - a média de temperatura pelas 4 estações do ano
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv"

dados = pd.read_csv(url)

ubiretama = dados[dados["nome_municipio"] == "Ubiretama"]

print("INFORMAÇÕES DE UBIRETAMA:")
print(ubiretama.to_string(index=False))


meses = ["jan", "fev", "mar", "abr", "mai", "jun",
         "jul", "ago", "set", "out", "nov", "dez"]

nomes_meses = {
    "jan": "Janeiro",
    "fev": "Fevereiro",
    "mar": "Março",
    "abr": "Abril",
    "mai": "Maio",
    "jun": "Junho",
    "jul": "Julho",
    "ago": "Agosto",
    "set": "Setembro",
    "out": "Outubro",
    "nov": "Novembro",
    "dez": "Dezembro"
}



temperaturas = ubiretama[
    [f"temp_media_{mes}" for mes in meses]
].iloc[0]

mes_temp_min = temperaturas.idxmin()
mes_temp_min = mes_temp_min.replace("temp_media_", "")

print("\nMês de menor temperatura média:")
print(nomes_meses[mes_temp_min])
print(f"Temperatura: {temperaturas.min():.2f} °C")



deficit = ubiretama[
    [f"def_{mes}" for mes in meses]
].iloc[0]

mes_def_max = deficit.idxmax()
mes_def_max = mes_def_max.replace("def_", "")

print("\nMês de maior déficit hídrico:")
print(nomes_meses[mes_def_max])
print(f"Déficit hídrico: {deficit.max():.2f} mm")

verao = ["dez", "jan", "fev"]
outono = ["mar", "abr", "mai"]
inverno = ["jun", "jul", "ago"]
primavera = ["set", "out", "nov"]

estacoes = {
    "Verão": verao,
    "Outono": outono,
    "Inverno": inverno,
    "Primavera": primavera
}

print("\nMÉDIA DE TEMPERATURA POR ESTAÇÃO:")

for estacao, meses_estacao in estacoes.items():
    valores = [
        ubiretama[f"temp_media_{mes}"].iloc[0]
        for mes in meses_estacao
    ]

    media = sum(valores) / len(valores)

    print(f"{estacao}: {media:.2f} °C")
```
