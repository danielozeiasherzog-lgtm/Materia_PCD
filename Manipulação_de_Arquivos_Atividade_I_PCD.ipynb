```python
#Atividade 1
def lerPersonagem():
    arquivo = open("personagem.txt", "r")

    for linha in arquivo:
        print(linha, end="")

    arquivo.close()


arquivo = open("personagem.txt", "w")

personagem = str(input("Informe o nome do seu personagem: "))
idade = int(input("Informe a idade do seu personagem: "))
classe = str(input("Informe a classe do seu personagem: "))
forca = int(input("Informe a força do seu personagem (0-10): "))
vida = int(input("Informe a vida do seu personagem (0-10): "))

arquivo.write("Nome: " + personagem + "\n")
arquivo.write("Idade: " + str(idade) + "\n")
arquivo.write("Classe: " + classe + "\n")
arquivo.write("Força: " + str(forca) + "\n")
arquivo.write("Vida: " + str(vida) + "\n")

arquivo.close()

print("\nPersonagem salvo!\n")

print("Dados do personagem:")
lerPersonagem()


```
